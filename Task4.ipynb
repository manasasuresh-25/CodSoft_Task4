```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df =pd.read_csv("customer_data.csv")
```

```python
df.info()
```

```python
df.duplicated().sum()
df.describe()
```

```python
# Create Age Groups

bins = [0, 25, 35, 45, 55, 100]
labels = ['18-25', '26-35', '36-45', '46-55', '56+']

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)

print(df['age_group'].value_counts().sort_index())
```

```python
# Spending by Age Group

age_spending = df.groupby(
    'age_group',
    observed=True
)['purchase_amount'].agg(['mean', 'sum'])

age_spending
```
