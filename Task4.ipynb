```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df =pd.read_csv("customer_data.csv")
```

```python
df.info()
```

```python
df.duplicated().sum()
df.describe()
```

```python
# Create Age Groups

bins = [0, 25, 35, 45, 55, 100]
labels = ['18-25', '26-35', '36-45', '46-55', '56+']

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)

print(df['age_group'].value_counts().sort_index())
```

```python
# Spending by Age Group

age_spending = df.groupby(
    'age_group',
    observed=True
)['purchase_amount'].agg(['mean', 'sum'])

age_spending
```

```python


frequency_counts = df['purchase_frequency'].value_counts()

print("Purchase Frequency:")
print(frequency_counts)
```

```python


frequency_spending = df.groupby(
    'purchase_frequency'
)['purchase_amount'].agg(['count', 'mean', 'sum'])

frequency_spending
```

```python
# most purchased data

category_counts = df['product_category'].value_counts()

print("Number of purchases by product category:")
print(category_counts)
```

```python
# Total and average spending by product category

category_spending = df.groupby(
    'product_category'
)['purchase_amount'].agg(['count', 'mean', 'sum'])

category_spending
```

```python
# Loyalty Status Analysis

loyalty_analysis = df.groupby(
    'loyalty_status'
)['purchase_amount'].agg(['count', 'mean', 'sum'])

loyalty_analysis
```

```python
# Promotion Usage Analysis

promotion_analysis = df.groupby(
    'promotion_usage'
)['purchase_amount'].agg(['count', 'mean', 'sum'])

promotion_analysis
```
